## 2. Galeri Visualisasi Laporan (300 DPI)

```
fig_paths = [
    '../reports/figures/sentiment_distribution.png',
    '../reports/figures/aspect_sentiment_breakdown.png',
    '../reports/figures/reputation_index_aspect.png',
    '../reports/figures/model_comparison_chart.png',
    '../reports/figures/wordcloud_combined.png',
]

for p in fig_paths:
    path = Path(p)
    if path.exists():
        img = mpimg.imread(str(path))
        plt.figure(figsize=(12, 6))
        plt.imshow(img)
        plt.axis('off')
        plt.title(path.name, fontsize=12, pad=10)
        plt.show()
    else:
        print(f"File belum ada: {path}")
```
