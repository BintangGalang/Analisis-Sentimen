# Eksplorasi Pelabelan & Validasi Inter-Annotator Agreement (Cohen's Kappa)
**Proyek:** Analisis Sentimen Reputasi Undiksha pada Media Sosial  
**Tahap:** 4 – Pelabelan & Validasi Ground Truth (PRD §6 F3)  


```
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Atur style visualisasi
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
```

## 1. Memuat Data Labeled & Evaluasi Double-Annotation

```
# Load data labeled final
labeled_path = Path('../data/labeled/youtube_labeled.csv')
eval_path = Path('../data/labeled/youtube_double_annotation_eval.csv')

df_labeled = pd.read_csv(labeled_path)
df_eval = pd.read_csv(eval_path)

print(f"Total data labeled: {len(df_labeled)} baris")
print(f"Total sampel double-annotation: {len(df_eval)} baris")
df_labeled.head(3)
```

## 2. Analisis Inter-Annotator Agreement (Cohen's Kappa)

```
from sklearn.metrics import cohen_kappa_score, confusion_matrix

labels = ['positif', 'netral', 'negatif']
kappa = cohen_kappa_score(df_eval['sentiment'], df_eval['sentiment_annotator_b'], labels=labels)
cm = confusion_matrix(df_eval['sentiment'], df_eval['sentiment_annotator_b'], labels=labels)

print(f"Cohen's Kappa Score : {kappa:.4f}")
print(f"Status Target (>=0.6): {'TERPENUHI (Substantial/Almost Perfect)' if kappa >= 0.6 else 'BELUM TERPENUHI'}")

cm_df = pd.DataFrame(cm, index=[f"Annotator A ({l})" for l in labels], columns=[f"Annotator B ({l})" for l in labels])
plt.figure(figsize=(6, 4))
sns.heatmap(cm_df, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix Antara Annotator A & Annotator B")
plt.tight_layout()
plt.show()
```

## 3. Distribusi Sentimen Ground Truth

```
dist = df_labeled['sentiment'].value_counts()
print(dist)

plt.figure(figsize=(7, 4))
colors = {'positif': '#2ecc71', 'netral': '#95a5a6', 'negatif': '#e74c3c'}
bar_colors = [colors.get(x, '#3498db') for x in dist.index]

plt.bar(dist.index, dist.values, color=bar_colors, edgecolor='black', alpha=0.85)
for i, v in enumerate(dist.values):
    plt.text(i, v + 5, f"{v} ({v/len(df_labeled)*100:.1f}%)", ha='center', fontweight='bold')

plt.title("Distribusi Kelas Sentimen (YouTube Undiksha)")
plt.xlabel("Sentimen")
plt.ylabel("Jumlah Komentar")
plt.ylim(0, max(dist.values) * 1.15)
plt.tight_layout()
plt.show()
```

## 4. Distribusi Aspek Reputasi Undiksha

```
aspect_dist = df_labeled['aspect'].value_counts()
print(aspect_dist)

plt.figure(figsize=(9, 4))
plt.barh(aspect_dist.index[::-1], aspect_dist.values[::-1], color='#3498db', edgecolor='black', alpha=0.85)
for i, v in enumerate(aspect_dist.values[::-1]):
    plt.text(v + 2, i, f"{v}", va='center', fontweight='bold')

plt.title("Distribusi Kategori Aspek Reputasi Undiksha")
plt.xlabel("Jumlah Komentar")
plt.ylabel("Aspek")
plt.tight_layout()
plt.show()
```
